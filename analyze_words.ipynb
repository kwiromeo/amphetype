```python
import os


workspace_dir = os.getcwd()
rel_path = "amphetype/lesson_builder/google_10000_words.txt"
word_path = os.path.join(workspace_dir, rel_path)

common_words = []
with open(word_path, "r") as f:
    common_words = f.readlines()

long_words = [word for word in common_words if len(word) > 10]

print(len(long_words))
```

```python
### Print long words
for word in long_words:
    print(word)
```
